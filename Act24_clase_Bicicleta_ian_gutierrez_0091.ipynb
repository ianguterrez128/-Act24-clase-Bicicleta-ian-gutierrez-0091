```python
class Bicicleta:
    def __init__(self, marca, tipo, velocidades, peso):
        self.marca = marca
        self.tipo = tipo
        self.velocidades = velocidades
        self.peso = peso
        self.llantas_infladas = False
        self.marcha_actual = 1 # Initialize current gear

    def check_mantenimiento(self):
        print(f"\n[CHECKLIST]: Revisando presión de llantas y cadena de la {self.marca}...")
        self.llantas_infladas = True
        print(f"[SISTEMA]: Presión óptima. ¡La bicicleta de {self.tipo} está lista para rodar!")

    def mostrar_especificaciones(self):
        print("\n" + "🚲  " * 5)
        print(f"FICHA TÉCNICA DE LA BICI")
        print("-" * 25)
        print(f"Marca:        {self.marca}")
        print(f"Tipo:         {self.tipo}")
        print(f"Velocidades:  {self.velocidades}")
        print(f"Peso:         {self.peso} kg")
        print("-" * 25)

    def pedalear(self):
        if not self.llantas_infladas:
            print(f"\n[!] No puedes pedalear, las llantas de la {self.marca} están desinfladas. ¡Por favor, ínflalas primero!")
        else:
            print(f"\n[MOVIMIENTO]: ¡Estás pedaleando en la marcha {self.marcha_actual}! La bicicleta de {self.tipo} está en marcha.")

    def cambiarMarcha(self, nueva_marcha):
        # Convertimos a int para asegurar la comparación numérica
        try:
            max_velocidades = int(self.velocidades)
            marcha_a_cambiar = int(nueva_marcha)
        except ValueError:
            print(f"\n[ERROR]: El número de velocidades '{self.velocidades}' o la marcha '{nueva_marcha}' no son valores numéricos válidos.")
            return

        if 1 <= marcha_a_cambiar <= max_velocidades:
            self.marcha_actual = marcha_a_cambiar
            print(f"\n[CAMBIO]: Marcha ajustada a nivel {self.marcha_actual}.")
        else:
            print(f"\n[ERROR]: La marcha {nueva_marcha} está fuera de rango (1-{max_velocidades}).")

    def inflarLlantas(self):
        self.llantas_infladas = True
        print(f"\n[AIRE]: Llantas infladas. Listas para rodar.")

# --- Inicio del simulador interactivo ---
print("TALLER DE CICLISMO - REGISTRO DE EQUIPO")
print("---------------------------------------")

# Interacción con el usuario (Entrada de datos)
mar = input("Marca de la bicicleta: ")
tip = input("Tipo (Montaña/Ruta/Urbana): ")
vel = input("Número de velocidades: ")
pes = input("Peso de la bicicleta (kg): ")

# Creación del objeto bicicleta
mi_bici = Bicicleta(mar, tip, vel, pes)

# Mostrar la información capturada
mi_bici.mostrar_especificaciones()

# Acción interactiva
ruta = input(f"\n¿Deseas cambiar marcha {mar}? (si/no): ").lower()

if ruta == "si":
    mi_bici.check_mantenimiento()
    print("🚴 ¡A pedalear! Disfruta el camino.")
else:
    print("❌ Ruta cancelada. La bicicleta se queda en el garage.")

print("Software de Control de Ciclistas finalizado")
print("Desarrollado por: Ian Gutiérrez | NL: 22")
```
